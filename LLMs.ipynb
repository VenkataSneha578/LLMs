What is an LLM? - Next token prediction Demo

```python
from transformers import pipeline

generator = pipeline("text-generation", model="distilgpt2")
prompt = "I love walking"
result = generator(prompt, max_new_tokens=10, num_return_sequences=3)
for i,output in enumerate(result,1):
  print(f"\nOutput {i}")
  print(output["generated_text"])
```

Tokens

```python
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained("gpt2")
text= "I love Hyderabad biryani!"
tokens = tokenizer.tokenize(text)
token_ids = tokenizer.encode(text)
print(f"Tokens: {tokens}")
print(f"Token IDs: {token_ids}")
print("\n Number of tokens:", len(tokens), "Number of IDs:", len(token_ids))
```

Converting tokens into embedding vectors

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
sentences = [
    "I love biryani",
    "Biryani is my favourite dish",
    "I love Hyderabad"
]
embeddings = model.encode(sentences)
for sentence, vector in zip(sentences, embeddings):
    print("Sentence:", sentence)
    print("Embedding size", len(vector))
    print("First 5 numbers:" , vector[:5])
```

```python
Why embeddings Matter? - similarity Demo
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
model= SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
sentences = [
    "I love biryani",
    "Biryani is my favourite dish",
    "I love Hyderabad",
    "Hyderabad is famous for food"
]
embeddings = model.encode(sentences)
query = "I like eating spicy rice food"
query_embedding = model.encode([query])
scores = cosine_similarity(query_embedding, embeddings)[0]
for sentence, score in zip(sentences, scores):
    print(f"{sentence} --> Similarity: {score:.3f}")
```
